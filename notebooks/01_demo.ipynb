# Démonstration portable
Installer le projet avec `%pip install -r /chemin/du/projet/requirements.txt` depuis la racine du clone (le `-e .` est relatif). Dans Colab, suivre d'abord les cellules du README. En local, sélectionner le Python de `.venv` dans votre outil notebook ; Jupyter/ipykernel restent optionnels et ne sont pas installés par le modèle.
Les chemins sont indépendants du dossier courant. Cette cellule utilise les données existantes, ou génère le petit CSV si absent.

```python
from scientific_project.config import load_config
from scientific_project.data import download_data
from scientific_project.paths import child_path, directories
from scientific_project.pipeline import run_pipeline
import json
config = load_config()
source = child_path(directories(config)[0] / 'raw', config['download']['filename'])
if not source.exists():
    download_data(config)
result = run_pipeline(config)
json.loads(result.read_text(encoding='utf-8'))
```
